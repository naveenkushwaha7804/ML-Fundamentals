```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler, RobustScaler
```

```python
df=pd.DataFrame({
    "Age":[18,22,30,45,60,80],
    "Salary":[20000,25000,35000,50000,70000,90000]
})
print("original Data :")
print(df)
```

```python
# MINMAX 
minmax_scaler=MinMaxScaler()
df_minmax=pd.DataFrame(minmax_scaler.fit_transform(df),
                      columns=df.columns)
print("\n Min Max Scaled Data:")
print(df_minmax)
```

```python
# STANDARD
standard_scaler=StandardScaler()
df_standard=pd.DataFrame(standard_scaler.fit_transform(df),columns=df.columns)
print("\n Standard Scaler Data :")
print(df_standard)
```

```python
# ROBUST
robust_scaler=RobustScaler()
df_robust=pd.DataFrame(robust_scaler.fit_transform(df),columns=df.columns)
print("\n Robust Scaler Data :")
print(df_robust)
```
