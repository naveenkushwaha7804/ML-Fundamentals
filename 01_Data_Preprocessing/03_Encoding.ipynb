```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
```

```python
df=pd.DataFrame({
    "Color":['Red','Blue','Green','blue','Red'],
    "Shape":['Circle','Square','Triangle','Circle','Square']
})
print("Original Data :")
print(df)
```

```python
# Label Encoding
le_Color=LabelEncoder()
le_Shape=LabelEncoder()
df['Color_encoded']=le_Color.fit_transform(df['Color'])
df['Shape_encoded']=le_Shape.fit_transform(df['Shape'])
print("\n Label Encoded Data :")
print(df)
```

```python
#Onehot
import pandas as pd
from sklearn.preprocessing import OneHotEncoder 

# Sample data
df=pd.DataFrame({
    "Color":['Red','Blue','Green','blue','Red'],
    "Shape":['Circle','Square','Triangle','Circle','Square']
})
print("Original Data :")
print(df)
```

```python
#2. One-hot encoding with sklearn
encoder = OneHotEncoder(sparse_output=False)
encoded = encoder.fit_transform(df[["Color","Shape"]])

#Create  Daataframe with feature names
df_encoded_sklearn=pd.DataFrame(encoded,columns=encoder.get_feature_names_out(["Color","Shape"]))
print("\nOne-Hot Encoded Data (sklearn):")
print(df_encoded_sklearn)
```

```python
# Ordinal Encoder 
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder
# Sample ordinal data
df=pd.DataFrame({
    "Size":["Small","Medium","Large","Medium","Small"],
    "Education":["Primary","Secondary","Graduate","PostGraduate","Graduate"]
})
print("Original Data :")
print(df)
```

```python
# Define order of categories
size_order=["Small","Medium","Large"]
education_order=["Primary","Secondary","Graduate","PostGraduate"]
```

```python
encoder=OrdinalEncoder(categories=[size_order,education_order])

df_encoded=df.copy()
df_encoded[["Size_encoded","Education_encoded"]]=encoder.fit_transform(df[["Size","Education"]])
print("\nOriginal data:")
print(df_encoded)
```
