```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.impute import SimpleImputer
import random
```

```python
# IMPUTATION
data={"id":[1,2,3,4,5,6],
     "age":[25,np.nan,30,35,np.nan,100],
     "salary":[50000,60000,np.nan,80000,90000,np.nan],
     "city":["New York","London",np.nan,"Parish","London",np.nan]
     }
df=pd.DataFrame(data)
print('Original Data :')
df
```

```python
# "Mean Imputer"
mean_imputer=SimpleImputer(strategy='mean')
df_mean=df.copy()
df_mean[['age','salary']]=mean_imputer.fit_transform(df_mean[['age','salary']])

print("\nMean Imputed Data:")
display(df_mean)
```

```python
# "Median Imputer"
median_imputer=SimpleImputer(strategy='median')
df_median=df.copy()
df_median[['age','salary']]=median_imputer.fit_transform(df_median[['age','salary']])

print("\nMedian Imputed Data:")
display(df_median)
```

```python
# Mode imputer
mode_imputer = SimpleImputer(strategy='most_frequent')
df_mode=df.copy()
df_mode[['age','salary','city']]=mode_imputer.fit_transform(df_mode[['age','salary','city']])
print("\nMode Imputed Data:")
display(df_mode)
```

```python
# Random Simpling Imputation
df_random=df.copy()

def random_impute(series):
    # drop nan value to avoid errors
    filled= series.dropna().tolist()
    return series.apply(lambda x:random.choice(filled) if pd.isnull(x) else x)

for col in ['age','salary','city']:
    df_random[col]=random_impute(df_random[col])

print('\nRandom Sampling Imputed Data:')
display(df_random)
```
